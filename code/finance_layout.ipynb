```python
import matplotlib.pyplot as plt
import networkx as nx
import pydot
from networkx.drawing.nx_pydot import graphviz_layout
```

```python
edge_list = (
    ("prod", "sales"),
    ("serv", "sales"),
    ("mktg", "opex"),
    ("gna", "opex"),
    ("tech", "opex"),
    ("fulf", "opex"),
    ("mktcap", "capebitda"),
    ("ebitda", "capebitda"),
    ("eq", "lev"),
    ("ta", "lev"),
    ("sales", "gp"),
    ("cogs", "gp"),
    ("sales", "turn"),
    ("ta", "turn"),
    ("gp", "ebit"),
    ("opex", "ebit"),
    ("ebit", "pretax"),
    ("nonop", "pretax"),
    ("abn", "pretax"),
    ("ebit", "rona"),
    ("na", "rona"),
    ("pretax", "ni"),
    ("taxexp", "ni"),
    ("ni", "ros"),
    ("sales", "ros"),
    ("ni", "roa"),
    ("ta", "roa"),
    ("ros", "roe"),
    ("turn", "roe"),
    ("lev", "roe"),
)
```

```python
G = nx.DiGraph()
for (u, v) in edge_list:
    G.add_edge(u, v)
```

```python
pos = nx.nx_pydot.graphviz_layout(G, prog="dot")   # the layered layout from last message

plt.figure(figsize=(13, 8))
nx.draw(
    G, pos,
    with_labels=True,
    node_color="#DCE6F1",
    node_size=1600,
    node_shape="s",
    font_size=7,
    arrows=True,
    arrowsize=12,
    edge_color="#666666",
)
plt.savefig("graph.png", dpi=150)
```
