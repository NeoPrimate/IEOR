```python
from scipy.stats import t, norm
import numpy as np

xbar = 10
s = 5
n = 1000

conf = 0.90

alpha = 1 - conf
p = 1 - alpha / 2
crit = t.ppf(p, df=n-1)
se = s / (np.sqrt(n))

print(f"Alpha: {alpha:.2f}")
print(f"Cumulative Probability: {p:.2f}")
print(f"Critical Value: {crit:.2f}")
print(f"Standard Error: {se:.2f}")

lower = xbar - crit * se
upper = xbar + crit * se

print(f"Confidence Interval: [{lower:.2f}, {upper:.2f}]")
```

```python
from scipy.stats import t
import numpy as np

xbar = 10
s = 5
n = 1000

conf = 0.99

p = 1 - ((1 - conf) / 2)
crit = t.ppf(p, df=n-1)

se_pred = s * np.sqrt(1 + (1 / n))

print(f"Standard Error Prediction: {se_pred:.2f}")

lower = xbar - crit * se_pred
upper = xbar + crit * se_pred

print(f"Prediction Interval: [{lower:.2f}, {upper:.2f}]")
```

```python
import numpy as np
from scipy.stats import norm, t

rng = np.random.default_rng(42)

mu, sigma = 0.0, 1.0   # properties of the MECHANISM — not of any data
n = 25                 # sample size — the only thing that sets precision
R = 10_000             # replications — only how clearly we can SEE the result
df = n-1
conf = 0.95

alpha = 1 - conf
p = 1 - alpha / 2

z_crit = norm.ppf(p)
t_crit = t.ppf(p, df=df)


def Xbar():            # ← the random variable X̄ (the estimator itself)
    return rng.normal(mu, sigma, n).mean()

# what you actually have in real life: one sample, R = 1
sample = rng.normal(mu, sigma, n)   # n realized observations
xbar   = sample.mean()              # your estimate — one number
s      = sample.std(ddof=1)         # your estimate of sigma

se = s / np.sqrt(n)

print(f"one observation   {sample[0]:.3f}")
print(f"xbar              {xbar:.3f}")
print(f"s                 {s:.3f}   estimates sigma = {sigma:.3f}")
print(f"s / sqrt(n)       {s / np.sqrt(n):.3f}")
print(f"ci (norm)         [{xbar - z_crit * se}, {xbar + z_crit * se}]")
print(f"ci (t)            [{xbar - t_crit * se}, {xbar + t_crit * se}]")

# the thought experiment you never get to run
xbars = np.array([Xbar() for _ in range(R)])

print()
print(f"spread of xbars   {xbars.std(ddof=1):.3f}   measured")
print(f"sigma / sqrt(n)   {sigma / np.sqrt(n):.3f}   derived, no simulation needed")
print(f"spread of data    {sample.std(ddof=1):.3f}   does not shrink with n")

print()
for n_ in (5, 25, 100, 1000):
    x = rng.normal(mu, sigma, size=(R, n_))
    print(f"n={n_:>5}   s={x.std(axis=1, ddof=1).mean():.3f}   se={x.mean(axis=1).std(ddof=1):.3f}")
```
