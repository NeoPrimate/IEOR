```python
import numpy as np
from scipy.stats import norm, expon

rng = np.random.default_rng(42)

sample = expon.rvs(loc=0, scale=1, size=10)
xbar = sample.mean()
s = sample.std()
n = sample.size

se = s / np.sqrt(n)
print(f'SE (closed): {se:.2f}')

B = 50_000
resamples = rng.choice(sample, size=(B, n), replace=True)
print(f"SE (bootstrap): {resamples.mean(axis=1).std(ddof=1):.2f}")
```
